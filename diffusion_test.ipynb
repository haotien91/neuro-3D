# one run

```python
import os
import glob
import shutil
import subprocess
import time
import re
import csv  # 用來存 CSV，不需要 pandas

BASE = "/content/drive/MyDrive/Google Colab/neuro-3D/neuro3d_hf"

# ====== 想跑哪些 subject & condition，直接改這裡 ======
SUBJECTS = ["sub03", "sub11"]
CONDITIONS = ["normal", "zero", "random"]
# =====================================================

# recon_main.py 的固定參數
GEN_TYPE = "shape"
TASK = "sample"
IN_CHANNELS = "1027"
PRETRAIN_PATTERN = "{sub}-72class-retri"
EXP_PATTERN = "{sub}-72class-shape"
CKPT_PATTERN = (
    "/content/drive/MyDrive/Google Colab/neuro-3D/neuro3d_hf/"
    "model/point_generate/shape/{sub}/{exp}/checkpoint-100000.pth"
)

# ground truth path (evaluate 用)
GT_DIR = f"{BASE}/point_cloud"

RESULTS = []  # 存所有 subject × condition 的結果


def parse_evaluate_output(text: str):
    """
    從 evaluate.py 的輸出抓 Chamfer / Precision / Recall / F1
    """
    metrics = {}

    patterns = {
        "CD": r"Chamfer Distance\s*:\s*([\d\.Ee+-]+)",
        "Precision": r"Precision.*?:\s*([\d\.Ee+-]+)",
        "Recall": r"Recall.*?:\s*([\d\.Ee+-]+)",
        "F1": r"F1 Score.*?:\s*([\d\.Ee+-]+)",
    }

    for key, pat in patterns.items():
        m = re.search(pat, text)
        metrics[key] = float(m.group(1)) if m else None

    return metrics


def run_one(sub: str, cond: str):
    print("=" * 80)
    print(f"🚀 Running subject={sub}, condition={cond}")
    print("=" * 80)

    exp_dir = EXP_PATTERN.format(sub=sub)
    ckpt_path = CKPT_PATTERN.format(sub=sub, exp=exp_dir)

    # --- Step 1: run recon_main.py ---
    print("🎬 Step 1: Running recon_main.py ...")
    subprocess.run(
        [
            "python",
            "recon_main.py",
            "--data_path",
            BASE + "/",
            "--generation_type",
            GEN_TYPE,
            "--task",
            TASK,
            "--sub",
            sub,
            "--in_channels",
            IN_CHANNELS,
            "--pretrain_model",
            PRETRAIN_PATTERN.format(sub=sub),
            "--checkpoint_resume",
            ckpt_path,
            "--condition_mode",
            cond,
        ],
        check=True,
    )

    # --- Step 2: find generated ply files ---
    src_dir = f"{BASE}/model/point_generate/shape/{sub}/{exp_dir}/checkpoint-100000"
    ply_files = glob.glob(os.path.join(src_dir, "*.ply"))

    if not ply_files:
        raise RuntimeError(
            f"⚠️ 在 {src_dir} 找不到任何 .ply，可能 recon_main 沒有寫入成功"
        )

    # --- Step 3: save to diffusion_results/sub/cond/run_id ---
    run_id = time.strftime("%Y%m%d_%H%M%S")
    dst_dir = f"{BASE}/diffusion_results/{sub}/{cond}/{run_id}"
    os.makedirs(dst_dir, exist_ok=True)

    print(f"📦 Step 2: Copying {len(ply_files)} .ply files to:\n{dst_dir}")
    for f in ply_files:
        shutil.copy2(f, dst_dir)

    # --- Step 4: Evaluate ---
    print("🧪 Step 3: Evaluating...")
    result = subprocess.run(
        [
            "python",
            "evaluate.py",
            "--pred_dir",
            dst_dir,
            "--gt_dir",
            GT_DIR,
            "--dist_thresh",
            "0.01",
        ],
        capture_output=True,
        text=True,
    )

    # 原始 evaluate 輸出還是照樣印出來
    print(result.stdout)
    if result.stderr:
        print("stderr:", result.stderr)

    # 解析分數
    metrics = parse_evaluate_output(result.stdout)
    metrics.update({"sub": sub, "cond": cond, "dir": dst_dir})

    RESULTS.append(metrics)

    print(f"🎉 Done: {sub} - {cond}\nOutput saved in:\n{dst_dir}\n")
```

```python
# ========== 一鍵跑完所有 subject × condition ==========
for sub in SUBJECTS:
    for cond in CONDITIONS:
        run_one(sub, cond)

# ---------- 印簡單的 summary ----------
print("\n=================== SUMMARY ===================\n")
if not RESULTS:
    print("⚠️ 沒有任何結果，可能中途出錯。")
else:
    # 固定欄位順序
    headers = ["sub", "cond", "2way_top1", "10way_top3", "CD", "F1", "dir"]
    # 印 header
    print("\t".join(headers))
    # 印每一列
    for r in RESULTS:
        row = [str(r.get(h, "")) for h in headers]
        print("\t".join(row))

    # ---------- 存成 CSV ----------
    os.makedirs(f"{BASE}/diffusion_results", exist_ok=True)
    csv_path = os.path.join(
        BASE,
        "diffusion_results",
        f"eval_summary_{time.strftime('%Y%m%d_%H%M%S')}.csv",
    )

    with open(csv_path, "w", newline="") as f:
        writer = csv.DictWriter(f, fieldnames=headers)
        writer.writeheader()
        for r in RESULTS:
            writer.writerow({h: r.get(h, "") for h in headers})

    print(f"\n💾 Summary CSV saved to:\n{csv_path}\n")
```
